```python
from google_play_scraper import Sort, reviews
import pandas as pd
import numpy as np
import time

wondrevs=[]
token = None

while len(wondrevs) < 16000:
  wonddata, token = reviews(
      'id.bni.wondr',
      lang='id',
      country='id',
      sort=Sort.NEWEST,
      count=200,
      filter_score_with=None,
      continuation_token=token
  )
  wondrevs.extend(wonddata)
  if token is None:
      break
  time.sleep(1)

wondcsv = pd.DataFrame(wondrevs)
wondcsv.to_csv('wond.csv', index = False, header=True)
```
